# Selecting the Number of Clusters from a Dendrogram

A dendrogram records every merge performed during agglomerative hierarchical clustering. Each horizontal joining bar marks a merge, and its vertical position represents the linkage dissimilarity or merge cost at which that merge occurred.

A common visual approach for selecting a candidate number of clusters is to identify the **largest vertical gap between consecutive merge levels**.

## The Largest Vertical-Gap Heuristic

To apply the method:

1. Identify the horizontal merge bars in the dendrogram.
2. Look for the largest vertical interval containing no horizontal merge.
3. Draw a horizontal threshold line through that interval.
4. Count the vertical branches crossed by the threshold.
5. Treat that count as a candidate number of clusters.

The intuition is that merges below the gap combine relatively similar observations, whereas the next merges above the gap require a substantially larger increase in dissimilarity. Cutting within the gap prevents those high-cost merges.

> Count the vertical branches intersected by the horizontal cut—not the horizontal merge bars and not the number of vertical segments visible inside the gap.

$$

## Why the Threshold Can Move Within the Gap

Any horizontal threshold placed strictly between the same two consecutive merge heights produces the same partition. Moving the line upward or downward inside that empty interval does not cross a merge, so no cluster membership changes.

The partition changes only when the threshold passes a horizontal merge bar:

- moving below a merge separates the two branches involved and increases the cluster count; and
- moving above a merge combines those branches and decreases the cluster count.

## A Mathematical View

For $n$ observations, agglomerative clustering performs $n-1$ merges. Let the ordered merge heights be:

$$
h_1\leq h_2\leq\cdots\leq h_{n-1}
$$

The gap after merge $j$ is:

$$
\Delta_j=h_{j+1}-h_j
$$

The largest-gap heuristic selects the index with the greatest $\Delta_j$ and places the threshold $t$ such that:

$$
h_j<t<h_{j+1}
$$

Because $j$ merges have been retained below the cut, the resulting number of clusters is:

$$
K=n-j
$$

This algebraic view is equivalent to counting the vertical branches intersected by the cut.

## What the Vertical Axis Means

The vertical axis represents the dissimilarity or merge cost defined by the selected metric and linkage criterion. It should not always be interpreted as an ordinary point-to-point Euclidean distance.

- Single, complete, and average linkage summarize pairwise distances in different ways.
- Centroid linkage measures separation between cluster means.
- Ward linkage represents the increase in within-cluster variation caused by a merge.

Consequently, the scale and meaning of a vertical gap depend on how the hierarchy was constructed.

# Study Notes

## Core idea

> Find a large merge-height gap, place a horizontal cut inside it, and count the vertical branches crossed to obtain a candidate number of clusters.

## Visual procedure

$$
\text{Merge heights}\rightarrow\text{largest empty vertical interval}\rightarrow\text{horizontal cut}\rightarrow\text{branch count}\rightarrow K
$$

## Important distinctions

| Concept | Meaning |
|---|---|
| Horizontal merge bar | Two clusters were joined at that height |
| Vertical gap | Interval between merge heights containing no merge |
| Horizontal threshold | Selected cut used to extract a partition |
| Vertical branch crossing | One cluster that exists below the cut |
| Large jump in height | A merge between groups that are much less similar under the chosen method |

## Threshold behavior

- A lower cut creates more, smaller clusters.
- A higher cut creates fewer, larger clusters.
- Cuts within the same merge-free interval produce the same clusters.
- A cut placed exactly at a merge height may require a tie-handling convention.

## Why “optimal” should be used carefully

The largest vertical gap gives a **candidate** value of $K$, not a guaranteed optimum. A clear visual gap can still reflect outliers, unsuitable scaling, or an inappropriate linkage rule.

Validate the candidate partition using:

- silhouette score or related internal measures;
- cluster sizes and profiles;
- stability across samples and time periods;
- sensitivity to scaling, metrics, and linkage;
- domain knowledge and interpretability; and
- usefulness for the intended business or research decision.

## Common mistakes

### Counting horizontal bars

The number of clusters is determined by the vertical branches crossed by the threshold.

### Searching only from the top

The goal is to compare all consecutive merge-height gaps. The largest useful gap may not always be the first one encountered from the top.

### Treating height as raw Euclidean distance in every model

The meaning of dendrogram height depends on the linkage criterion and metric.

### Assuming a large gap proves natural clusters

A large gap is evidence of separation under the chosen configuration, not proof of an objectively correct structure.

### Ignoring outliers

An isolated observation may merge at a very large height and create the largest gap, leading to one singleton cluster and one broad group. Investigate such cases before accepting the result.
